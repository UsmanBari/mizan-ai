# 04 · Results for the live demo
Collects everything from `eval/results/summary.json` into the tables and figures shown on the demo slides. The same JSON also feeds the web app's **Evaluation** page (`/eval`). Every number can be reproduced with the commands in each notebook.

```python
# --- Setup (works on Google Colab and locally) ---------------------------------
import os, sys, subprocess
IN_COLAB = "google.colab" in sys.modules
if IN_COLAB and not os.path.exists("mizan-ai"):
    subprocess.run(["git", "clone", "https://github.com/usmanbari-ub/mizan-ai.git"], check=True)
ROOT = os.path.abspath("mizan-ai" if IN_COLAB else "../..")
os.chdir(os.path.join(ROOT, "eval"))
sys.path[:0] = [os.path.join(ROOT, "eval"), os.path.join(ROOT, "backend")]
if IN_COLAB:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", os.path.join(ROOT, "backend/requirements.txt")], check=True)
print("working dir:", os.getcwd())
```

```python
import json, pandas as pd
S = json.load(open("results/summary.json"))
for section, data in S.items():
    print("=" * 20, section)
    data = {k: v for k, v in data.items() if isinstance(v, dict) and k != "paired_tests"}
    display(pd.DataFrame(data).T)
```

```python
# Functional test cases (product flows) - pass rate for the demo
import subprocess
r = subprocess.run([sys.executable, "-m", "pytest", "-q", os.path.join(ROOT, "backend/tests"), "-p", "no:cacheprovider"],
                   capture_output=True, text=True)
print(r.stdout[-1500:])
```
